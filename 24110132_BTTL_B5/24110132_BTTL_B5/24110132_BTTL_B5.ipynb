```python
class Node:
    def __init__(self, state, parent=None, action=None):
        self.state = state
        self.parent = parent
        self.action = action
```

```python
def Stack():
    frontier = []
    return frontier
```

```python
def push(frontier, node):
    frontier.append(node)
```

```python
def pop(frontier):
    return frontier.pop()
```

```python
def is_empty(frontier):
    return len(frontier) == 0
```

```python
def child_node(problem, node, action):
    return Node(problem.result(node.state, action), node, action)
```

```python
def solution(node):
    acts = []
    states = []
    while node is not None:
        states.insert(0, node.state)
        if node.action is not None:
            acts.insert(0, node.action)
        node = node.parent
    return acts, states
```

```python
class Problem:
    def __init__(self, initial_state, is_goal, actions, result):
        self.initial_state = initial_state
        self.is_goal = is_goal
        self.actions = actions
        self.result = result
```

```python
def DFS(problem):
    frontier = Stack()
    push(frontier, Node(problem.initial_state))
    explored = set()
    while not is_empty(frontier):
        node = pop(frontier)
        if problem.is_goal(node.state):
            return solution(node)
        explored.add(node.state)
        for action in problem.actions(node.state):
            child = child_node(problem, node, action)
            if child.state not in explored:
                push(frontier, child)
    return None
```

```python
# VACUUM ROBOT
def vacuum_is_goal(state):
    return state[1] == 'C' and state[2] == 'C'
```

```python
def vacuum_actions(state):
    pos, a, b = state
    acts = []
    if pos == 'A':
        acts.append('Right')
    else:
        acts.append('Left')
    if (pos == 'A' and a == 'D') or (pos == 'B' and b == 'D'):
        acts.append('Suck')
    return acts
```

```python
def vacuum_result(state, action):
    pos, a, b = state
    if action == 'Left':
        return ('A', a, b)
    if action == 'Right':
        return ('B', a, b)
    if pos == 'A':
        return ('A', 'C', b)
    return ('B', a, 'C')
```

```python
problem = Problem(('A', 'D', 'D'), vacuum_is_goal, vacuum_actions, vacuum_result)
kq = DFS(problem)

if kq is None:
    print("failure")
else:
    acts, states = kq
    print("So buoc:", len(acts))
    print("Hanh dong:", acts)
    for i in range(len(states)):
        pos, a, b = states[i]
        buoc = "Bat dau" if i == 0 else acts[i - 1]
        print(f"{i}. {buoc:8} -> robot o {pos}, A = {a}, B = {b}")
```

```python
# 8-PUZZLE
def puzzle_is_goal(state):
    return state == (1, 2, 3, 4, 5, 6, 7, 8, 0)
```

```python
def puzzle_actions(state):
    i = state.index(0)
    acts = []
    if i % 3 != 0:
        acts.append('Left')
    if i % 3 != 2:
        acts.append('Right')
    if i >= 3:
        acts.append('Up')
    if i <= 5:
        acts.append('Down')
    return acts
```

```python
def puzzle_result(state, action):
    step = {'Left': -1, 'Right': 1, 'Up': -3, 'Down': 3}
    s = list(state)
    i = s.index(0)
    j = i + step[action]
    s[i], s[j] = s[j], s[i]
    return tuple(s)
```

```python
def print_board(state):
    print(state[0:3])
    print(state[3:6])
    print(state[6:9])
    print()
```

```python
problem = Problem((1, 2, 3, 4, 0, 6, 7, 5, 8), puzzle_is_goal, puzzle_actions, puzzle_result)
kq = DFS(problem)

if kq is None:
    print("failure")
else:
    acts, states = kq
    print("So buoc:", len(acts))
    print("Hanh dong:", acts)
    for i in range(len(states)):
        print("Bat dau" if i == 0 else f"Buoc {i}: {acts[i - 1]}")
        print_board(states[i])
```
