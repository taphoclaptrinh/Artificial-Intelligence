```python
class Node:
    def __init__(self, state, parent=None, action=None, path_cost=0):
        self.state = state
        self.parent = parent
        self.action = action
        self.path_cost = path_cost
```

```python
def PriorityQueue():
    frontier = []
    return frontier
```

```python
def enqueue(frontier, node):
    frontier.append(node)
    frontier.sort(key=lambda n: n.path_cost)
```

```python
def dequeue(frontier):
    return frontier.pop(0)
```

```python
def is_empty(frontier):
    return len(frontier) == 0
```

```python
def find_in_frontier(frontier, state):
    for i, node in enumerate(frontier):
        if node.state == state:
            return i
    return -1
```

```python
def child_node(problem, node, action):
    next_state = problem.result(node.state, action)
    cost = node.path_cost + problem.step_cost(node.state, action, next_state)
    return Node(next_state, node, action, cost)
```

```python
def solution(node):
    acts = []
    states = []
    total_cost = node.path_cost
    while node is not None:
        states.insert(0, node.state)
        if node.action is not None:
            acts.insert(0, node.action)
        node = node.parent
    return acts, states, total_cost
```

```python
class Problem:
    def __init__(self, initial_state, is_goal, actions, result, step_cost):
        self.initial_state = initial_state
        self.is_goal = is_goal
        self.actions = actions
        self.result = result
        self.step_cost = step_cost
```

```python
def UCS(problem):
    node = Node(problem.initial_state, path_cost=0)
    frontier = PriorityQueue()
    enqueue(frontier, node)
    explored = set()
    while not is_empty(frontier):
        node = dequeue(frontier)
        if problem.is_goal(node.state):
            return solution(node)
        explored.add(node.state)
        for action in problem.actions(node.state):
            child = child_node(problem, node, action)
            idx = find_in_frontier(frontier, child.state)
            if child.state not in explored and idx == -1:
                enqueue(frontier, child)
            elif idx != -1 and frontier[idx].path_cost > child.path_cost:
                frontier[idx] = child
                frontier.sort(key=lambda n: n.path_cost)
    return None
```

```python
# VACUUM ROBOT
def vacuum_is_goal(state):
    return state[1] == 'C' and state[2] == 'C'
```

```python
def vacuum_actions(state):
    pos, a, b = state
    acts = []
    if (pos == 'A' and a == 'D') or (pos == 'B' and b == 'D'):
        acts.append('Suck')
    if pos == 'A':
        acts.append('Right')
    else:
        acts.append('Left')
    return acts
```

```python
def vacuum_result(state, action):
    pos, a, b = state
    if action == 'Left':
        return ('A', a, b)
    if action == 'Right':
        return ('B', a, b)
    if pos == 'A':
        return ('A', 'C', b)
    return ('B', a, 'C')
```

```python
def vacuum_step_cost(state, action, next_state):
    return 1 if action == 'Suck' else 2
```

```python
problem = Problem(('A', 'D', 'D'), vacuum_is_goal, vacuum_actions, vacuum_result, vacuum_step_cost)
kq = UCS(problem)

if kq is None:
    print("failure")
else:
    acts, states, total_cost = kq
    print("So buoc:", len(acts))
    print("Tong chi phi:", total_cost)
    print("Hanh dong:", acts)
    for i in range(len(states)):
        pos, a, b = states[i]
        buoc = "Bat dau" if i == 0 else acts[i - 1]
        print(f"{i}. {buoc:8} -> robot o {pos}, A = {a}, B = {b}")
```

```python
# 8-PUZZLE
def puzzle_is_goal(state):
    return state == (1, 2, 3, 4, 5, 6, 7, 8, 0)
```

```python
def puzzle_actions(state):
    i = state.index(0)
    acts = []
    if i % 3 != 0:
        acts.append('Left')
    if i % 3 != 2:
        acts.append('Right')
    if i >= 3:
        acts.append('Up')
    if i <= 5:
        acts.append('Down')
    return acts
```

```python
def puzzle_result(state, action):
    step = {'Left': -1, 'Right': 1, 'Up': -3, 'Down': 3}
    s = list(state)
    i = s.index(0)
    j = i + step[action]
    s[i], s[j] = s[j], s[i]
    return tuple(s)
```

```python
def puzzle_step_cost(state, action, next_state):
    return 1 if action in ('Up', 'Down') else 2
```

```python
def print_board(state):
    print(state[0:3])
    print(state[3:6])
    print(state[6:9])
    print()
```

```python
problem = Problem((1, 2, 3, 4, 0, 6, 7, 5, 8), puzzle_is_goal, puzzle_actions, puzzle_result, puzzle_step_cost)
kq = UCS(problem)

if kq is None:
    print("failure")
else:
    acts, states, total_cost = kq
    print("So buoc:", len(acts))
    print("Tong chi phi:", total_cost)
    print("Hanh dong:", acts)
    for i in range(len(states)):
        print("Bat dau" if i == 0 else f"Buoc {i}: {acts[i - 1]}")
        print_board(states[i])
```
